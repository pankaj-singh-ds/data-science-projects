```python
df = pd.get_dummies(df,columns=['department','role_level','attrition'])
```

```python
df.isnull().sum()
```

```python
y = df["monthly_salary"]
x = df.drop(["absences_days","avg_weekly_hours"], axis =1)
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = 0.2, random_state = 42 )
 
```

```python
from sklearn.ensemble import RandomForestClassifier

model3 = RandomForestClassifier(
    n_estimators = 100,
    max_depth = 7,
    random_state = 42
)

model3.fit(x_train, y_train)

print("Random Forest Score:", model3.score(x_test, y_test))
```

```python
from sklearn.tree import DecisionTreeRegressor

model2 = DecisionTreeRegressor()

model2.fit(x_train, y_train)
print("Dicision Tree Score:", model2.score(x_test, y_test))
```
